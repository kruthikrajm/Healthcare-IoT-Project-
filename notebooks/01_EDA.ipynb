```python
import pandas as pd
import numpy as np

#MININET CREATION FINAL CODE
# =====================================================
# LOAD MININET ZEEK DATASET
# =====================================================

df = pd.read_csv(
    "../data/mininet_zeek_dataset.csv"
)


print("Original Shape:")
print(df.shape)


print(df.head())



# =====================================================
# RENAME ZEEK COLUMNS TO TON-IoT FORMAT
# =====================================================


rename_columns = {

    "id.orig_h":"src_ip",
    "id.resp_h":"dst_ip",

    "orig_bytes":"src_bytes",
    "resp_bytes":"dst_bytes",

}


df.rename(
    columns=rename_columns,
    inplace=True
)



# =====================================================
# ADD REQUIRED FEATURES
# =====================================================


required_features = [

"src_ip",
"src_port",
"dst_ip",
"dst_port",

"proto",
"service",
"duration",

"src_bytes",
"dst_bytes",

"conn_state",

"missed_bytes",

"src_pkts",
"src_ip_bytes",

"dst_pkts",
"dst_ip_bytes",


"dns_query",
"dns_qclass",
"dns_qtype",
"dns_rcode",

"dns_AA",
"dns_RD",
"dns_RA",
"dns_rejected",


"ssl_version",
"ssl_cipher",
"ssl_resumed",
"ssl_established",


"http_trans_depth",
"http_method",
"http_uri",
"http_version",

"http_request_body_len",
"http_response_body_len",
"http_status_code",

"http_user_agent",

"http_orig_mime_types",
"http_resp_mime_types",


"weird_name",
"weird_addl",
"weird_notice",


"type"

]



for col in required_features:

    if col not in df.columns:

        df[col]="unknown"



# =====================================================
# NUMERIC FEATURES
# =====================================================


numeric_features=[

"src_port",
"dst_port",

"duration",

"src_bytes",
"dst_bytes",

"missed_bytes",

"src_pkts",
"src_ip_bytes",

"dst_pkts",
"dst_ip_bytes",

"dns_qclass",
"dns_qtype",
"dns_rcode",

"http_trans_depth",

"http_request_body_len",
"http_response_body_len",

"http_status_code"

]



for col in numeric_features:

    df[col]=pd.to_numeric(

        df[col],

        errors="coerce"

    )

    df[col]=df[col].fillna(0)



# =====================================================
# CATEGORICAL FEATURES
# =====================================================


categorical_features=[

x for x in required_features

if x not in numeric_features

]



for col in categorical_features:


    df[col]=df[col].astype(str)


    df[col]=df[col].replace(

        "nan",

        "unknown"

    )


    df[col]=df[col].replace(

        "",

        "unknown"

    )



# =====================================================
# KEEP ONLY TRAINING FEATURES
# =====================================================


final_columns=[

'proto',
'service',
'duration',
'src_bytes',
'dst_bytes',
'conn_state',
'missed_bytes',
'src_pkts',
'src_ip_bytes',
'dst_pkts',
'dst_ip_bytes',

'dns_query',
'dns_qclass',
'dns_qtype',
'dns_rcode',

'dns_AA',
'dns_RD',
'dns_RA',
'dns_rejected',

'ssl_version',
'ssl_cipher',
'ssl_resumed',
'ssl_established',

'http_trans_depth',
'http_method',
'http_uri',
'http_version',

'http_request_body_len',
'http_response_body_len',
'http_status_code',

'http_orig_mime_types',
'http_resp_mime_types',

'weird_name',
'weird_addl',
'weird_notice',

'type'

]



df=df[final_columns]



# =====================================================
# SAVE FINAL DATASET
# =====================================================


df.to_csv(

    "mininet_processed.csv",

    index=False

)



print("\nConversion Completed")

print("========================")

print("Final Shape:")

print(df.shape)


print("\nClass Distribution:")

print(df["type"].value_counts())


print("\nSaved File:")

print("mininet_processed.csv")
```
